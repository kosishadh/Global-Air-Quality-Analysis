# 10  most polluted cities

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv("data.csv")
avg_pm2 = df.groupby("City")["PM2.5"].mean()
top_10_city = avg_pm2.sort_values(ascending=False).head(10)
print(top_10_city)
top_10_city.plot(kind='barh', color="Red", edgecolor = "black")
plt.xlabel('Average PM 2.5')
plt.show()
```

# Highest Average Pollution

```python
import matplotlib.pyplot as plt
import pandas as pd
df = pd.read_csv('data.csv')
avg_pm2 = df.groupby('Country')['PM2.5'].mean()
top_country = avg_pm2.sort_values(ascending=False).head(10)
print(top_country)
top_country.plot(kind='barh')
plt.xlabel('PM2.5')
plt.show()
```
